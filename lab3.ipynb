```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
df=pd.read_csv(r"C:\Users\Admin\Downloads\archive (4)\Titanic-Dataset.csv")
```

```python
df.head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"],bins=20,kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Survived",data=df)
plt.title("Survival Count")
plt.xlabel("Survived (0=No,1=Yes)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["Age"])
plt.title("Age BoxPlot")
plt.show()
```

```python
age = sorted(df["Age"].dropna())

n = len(age)

Q1 = age[n // 4]
Q3 = age[(3 * n) // 4]

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Age"] < lower) | (df["Age"] > upper)]

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Limit =", lower)
print("Upper Limit =", upper)
print("Number of Outliers =", len(outliers))

print(outliers)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Sex", data=df)

plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.barplot(data=df, x="Pclass", y="Fare", palette="viridis")
plt.title("Average Frae by Passenger Class")
plt.show()
```
